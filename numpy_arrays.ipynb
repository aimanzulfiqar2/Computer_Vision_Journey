```python
import numpy as np 
```

```python
my_list=[1,2,3,4]
print(type(my_list))
```

```python
#cast list as array
arr=np.array(my_list)
print(arr)
print(type(arr))
```

```python
#create an array using range function 
my_arr=np.arange(0,11)
print(my_arr)
my_arr2=np.arange(0,11,2)
print(my_arr2)
```

```python
#create zeros and ones array and it returns float values

zeros_arr=np.zeros((3,6))
print(zeros_arr)

ones_arr=np.ones((3,6))
print(ones_arr)
```

```python
#get same random value everytime using seed

np.random.seed(101)
rand_arr=np.random.randint(0,100,10)
print(rand_arr)


rand_arr2=np.random.randint(0,100,10)
print(rand_arr2)

```
